```python
import duckdb
import pandas as pd
from google.colab import files

clean_df1 = duckdb.query("""
SELECT
    seating_no,
    arabic_name,
    CASE
        WHEN total_degree = 0 AND student_case_desc = 'ناجح دور أول' THEN NULL
        ELSE total_degree
    END AS total_degree,
    student_case_desc
FROM df1
""").df()

clean_df1.to_excel('cleaned_thanawya_system.xlsx', index=False)
files.download('cleaned_thanawya_system.xlsx')
clean_df2 = duckdb.query("""
SELECT
    *,
    CASE
        WHEN total_degree = 0 AND student_case_desc = 'ناجح دور ثان' THEN NULL
        ELSE total_degree
    END AS total_degree_cleaned
FROM df2
""").df()
clean_df2['total_degree'] = clean_df2['total_degree_cleaned']
clean_df2 = clean_df2.drop(columns=['total_degree_cleaned'])

clean_df2.to_excel('cleaned_stage_new_search.xlsx', index=False)
files.download('cleaned_stage_new_search.xlsx')
```
